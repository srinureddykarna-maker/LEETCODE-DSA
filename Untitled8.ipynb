```python
x = torch.tensor(2.)

x.requires_grad_(True)

y = x ** 2

y.backward()

print(x.grad)
```

```python
import torch

# Input
x = torch.tensor(2., requires_grad=True)

# 1st operation
y = x * 3

# 2nd operation
z = y ** 2

# Before backward
print("x =", x)
print("y =", y)
print("z =", z)
print("x.grad before backward =", x.grad)

# Calculate gradients
z.backward()

# After backward
print("x.grad after backward =", x.grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)

with torch.no_grad():
    a = x * 3

print(a.requires_grad)   # False

b = x * 4

print(b.requires_grad)   # True
```
